# Test Data Prediction Multi

## Importing Libraries

```python
import numpy as np 
import pandas as pd
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import joblib
```

## Scaled Data

```python
df = pd.read_csv(r'C:\Users\FARJAD FARHAT\Downloads\MLProject\Phase3\TrainModelDataFinalScl.csv')
```

```python
def categorize_delay(delay):
    if delay == 0:
        return 'No Delay'
    elif delay < 0.3:
        return 'Short Delay'
    elif 0.3 <= delay <= 0.90:
        return 'Moderate Delay'
    else:
        return 'Long Delay'
```

```python
y = df['departure_delay'].apply(categorize_delay)
X = df.drop(columns=['departure_delay'])
```

```python
dfTest = pd.read_csv('TestModelDataFinal.csv')
```

```python
print(len(X.columns),len(dfTest.columns))
```

```python
model = LogisticRegression(C = 10, max_iter = 500, penalty = 'l2', solver = 'lbfgs')
model.fit(X,y)
```

```python
y_pred = model.predict(dfTest)
```

```python
data = pd.DataFrame(
    {
        "ID" : [i for i in range(1 , len(y_pred)+1)],
        "Delay" : y_pred
    }
)
```

```python
data.to_csv('PredictionsScaledMulti.csv', index=False)
```

## UnScaled Data

```python
df = pd.read_csv(r'C:\Users\FARJAD FARHAT\Downloads\MLProject\Phase3\TrainModelDataFinalNotScl.csv')
```

```python
def categorize_delay(delay):
    if delay == 0:
        return 'No Delay'
    elif delay < 45:
        return 'Short Delay'
    elif 45 <= delay <= 175:
        return 'Moderate Delay'
    else:
        return 'Long Delay'
```

```python
y = df['departure_delay'].apply(categorize_delay)
X = df.drop(columns=['departure_delay'])
```

```python
dfTest = pd.read_csv('TestModelDataFinal.csv')
```

```python
print(len(X.columns),len(dfTest.columns))
```

```python
model = LogisticRegression(C = 0.001, max_iter = 1000, penalty = 'l2', solver = 'lbfgs')
model.fit(X,y)
```

```python
y_pred = model.predict(dfTest)
```

```python
data = pd.DataFrame(
    {
        "ID" : [i for i in range(1 , len(y_pred)+1)],
        "Delay" : y_pred
    }
)
```

```python
data.to_csv('PredictionsNotScaledMulti.csv', index=False)
```
