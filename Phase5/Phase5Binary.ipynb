## Scaled Data

```python
df = pd.read_csv(r'C:\Users\FARJAD FARHAT\Downloads\MLProject\Phase3\TrainModelDataFinalScl.csv')
```

```python
y = df['departure_delay'].apply(lambda x: 'on-time' if x == 0 else 'delayed')
X = df.drop(columns=['departure_delay'])
```

```python
dfTest = pd.read_csv('TestModelDataFinal.csv')
```

```python
print(len(X.columns),len(dfTest.columns))
```

#### Random Forest

```python
model = RandomForestClassifier(random_state=42, class_weight='balanced',max_depth = None, min_samples_leaf = 2, min_samples_split = 2, n_estimators = 200)
model.fit(X,y)
```

```python
y_pred = model.predict(dfTest)
```

```python
data = pd.DataFrame(
    {
        "ID" : [i for i in range(1 , len(y_pred)+1)],
        "Delay" : y_pred
    }
)
```

```python
data.to_csv('PredictionsScaled.csv', index=False)
```

#### Logistic Regression

```python
model = LogisticRegression(max_iter=1000, C = 10, solver ='liblinear')
model.fit(X,y)
```

```python
y_pred = model.predict(dfTest)
```

```python
data = pd.DataFrame(
    {
        "ID" : [i for i in range(1 , len(y_pred)+1)],
        "Delay" : y_pred
    }
)
```

```python
data.to_csv('PredictionsScaledLr.csv', index=False)
```
