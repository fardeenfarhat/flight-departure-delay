# One Hot Encoding 

## Importing Libraries 

```python
import numpy as np 
import pandas as pd
```

### Using Scaled Data

```python
df = pd.read_csv(r"C:\Users\FARJAD FARHAT\Downloads\MLProject\Phase1\scaledDelay.csv")
```

```python
df.head(5)
```

```python
categorical_cols = ['Type', 'Status','Departure IATA Code', 'Departure ICAO Code','Days', 'Month', 'Year', 'Hour']
df[categorical_cols] = df[categorical_cols].fillna('unknown')
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

```python
features = ['Temperature (°F)avg','Temperature (°F)Max','Temperature (°F)Min','Humidity (%)avg','Humidity (%)Max','Humidity (%)Min',
'Wind Speed (mph)avg','Wind Speed (mph)Max','Wind Speed (mph)Min', 'Pressure (in)avg', 'Pressure (in)Max', 'Pressure (in)Min',
'Dew Point (°F)Max', 'Dew Point (°F)avg', 'Dew Point (°F)Min'
] + [col for col in df.columns if 'Departure IATA Code' in col or 'Departure ICAO Code' in col or 'Month' in col or 'Year' in col or 'status'
in col or 'Days' in col or 'Hour' in col]

X = df[features]  
y = df['departure_delay']   
```

```python
temp = df[features + ['departure_delay']]
```

```python
temp.to_csv('TrainModelDataFinalScl.csv', index=False)
```
