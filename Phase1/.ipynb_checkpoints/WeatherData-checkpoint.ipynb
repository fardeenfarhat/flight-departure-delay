```python
import numpy as np 
import pandas as pd 
import os
```

```python
def get_filenames(directory):
    filenames = os.listdir(directory)    
    return filenames
```

```python
basePath = r'C:\Users\FARJAD FARHAT\Downloads\MLProject\ML-Proj-Dataset\Weather'
names = get_filenames(basePath)
```

```python
names = sorted(names, key=lambda x: int(x.split('.')[0]))
names
```

```python
files = []
for i in names:
    temp = pd.read_excel(os.path.join(basePath,i))
    files.append(temp)
```

```python
def single_ones(data):
    temp = data[0].split()
    return temp[0],temp[1:]
```

```python
def multiple_ones(data):
    temp = data[0].split()
    max_temp = []
    avg_temp = []
    min_temp = []
    for i in range(3,len(temp),3):
        max_temp.append(temp[i])
        avg_temp.append(temp[i+1])
        min_temp.append(temp[i+2])
    return max_temp,avg_temp,min_temp
```

```python
mult = ["Temperature (°F)","Dew Point (°F)","Humidity (%)","Wind Speed (mph)","Pressure (in)"]
year = ["2023","2024"]
months = []
months_year = []
vals = ["Max","avg","Min"]
dictionaries = []
for i in files:
    dicti = {}
    print(i["Time"])
    _,days = single_ones(i["Time"].tolist())
    if len(months) >= 6:
        months.append(_+year[1])
        months_year.append(year[1])
    else:
        months.append(_)
        months_year.append(year[0])
    dicti["Days"] = days
        
    for j in mult:
        a,b,c = multiple_ones(i[j].tolist())
        dicti[j+vals[0]] = a
        dicti[j+vals[1]] = b
        dicti[j+vals[2]] = c
        
    dictionaries.append(dicti)
```

```python
os.makedirs("Weather_Months", exist_ok=True)
path = r'C:\Users\FARJAD FARHAT\Downloads\MLProject\Phase1\Weather_Months'
```

```python
for i in range(len(dictionaries)):
    fileName = months[i]+'.csv'
    newFileName = os.path.join(path, fileName)
    df = pd.DataFrame(dictionaries[i])
    df.to_csv(newFileName,index=False)
```
