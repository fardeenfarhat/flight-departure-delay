```python
from docx import Document
import numpy as np
import pandas as pd 
import json
import re
import os
```

## NEW TEST DATA

```python
df = pd.read_csv('Binary_Test_File.csv')
```

```python
df.drop(columns=['File Name'],inplace=True)
df
```

```python
columns = ['Scheduled Time','Arrival Estimated Time']
for col in columns:
    df[col] = pd.to_datetime(df[col], format='%Y-%m-%dT%H:%M:%S.%f', errors='coerce')
```

```python
df['Days'] = df['Scheduled Time'].dt.day
df['Month'] = df['Scheduled Time'].dt.strftime('%b')
df['Year'] = df['Scheduled Time'].dt.year.astype(str)
df['Hour'] = df['Scheduled Time'].dt.hour
df
```

```python
df.to_csv('Test.csv',index=False)
```
