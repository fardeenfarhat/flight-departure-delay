```python
from docx import Document
import numpy as np
import pandas as pd 
import json
import re
import os
```

```python
def extractText(filename):
    doc = Document(filename)
    text = ""
    for paragraph in doc.paragraphs:
        text += paragraph.text.strip()
    cleaned_text = text.replace('\n', '').replace('\t', '')
    return json.loads(text)

def get_filenames(directory):
    filenames = os.listdir(directory)    
    return filenames
```

```python
def flatten_and_convert_to_dataframe(data):
    flattened_data = []
    for entry in data:
        flat_entry = {
            "type": entry["type"],
            "status": entry["status"],
            "departure_iataCode": entry["departure"].get("iataCode", np.nan),
            "departure_icaoCode": entry["departure"].get("icaoCode", np.nan),
            "departure_terminal": entry["departure"].get("terminal", np.nan),
            "departure_scheduledTime": entry["departure"].get("scheduledTime", np.nan),
            "departure_actualTime": entry["departure"].get("actualTime", np.nan),
            "arrival_iataCode": entry["arrival"].get("iataCode", np.nan),
            "arrival_icaoCode": entry["arrival"].get("icaoCode", np.nan),
            "arrival_terminal": entry["arrival"].get("terminal", np.nan),
            "arrival_scheduledTime": entry["arrival"].get("scheduledTime", np.nan),
            "arrival_estimatedTime": entry["arrival"].get("estimatedTime", np.nan),
            "airline_name": entry["airline"].get("name", np.nan),
            "flight_number": entry["flight"].get("number", np.nan),
            "flight_iataNumber": entry["flight"].get("iataNumber", np.nan),
            "flight_icaoNumber": entry["flight"].get("icaoNumber", np.nan),
            "codeshared_airline_name": entry.get("codeshared", {}).get("airline", {}).get("name", np.nan),
            "codeshared_airline_iataCode": entry.get("codeshared", {}).get("airline", {}).get("iataCode", np.nan),
            "codeshared_airline_icaoCode": entry.get("codeshared", {}).get("airline", {}).get("icaoCode", np.nan),
            "codeshared_flight_number": entry.get("codeshared", {}).get("flight", {}).get("number", np.nan),
            "codeshared_flight_iataNumber": entry.get("codeshared", {}).get("flight", {}).get("iataNumber", np.nan),
            "codeshared_flight_icaoNumber": entry.get("codeshared", {}).get("flight", {}).get("icaoNumber", np.nan)
        }
        flattened_data.append(flat_entry)
    return pd.DataFrame(flattened_data)
```

```python
basePath = r'C:\Users\FARJAD FARHAT\Downloads\MLProject\ML-Proj-Dataset\Test'

names = get_filenames(basePath)
files = []
for i in names:
    print(i)
    out = extractText(os.path.join(basePath,i))
    files.append(out)
```

```python
dfs = [] 
for file in files :
    temp = flatten_and_convert_to_dataframe(file)
    dfs.append(temp)
```

```python
df = pd.concat(dfs,axis=0,ignore_index=True)
df
```

```python
df = df.replace('',np.nan , regex=True)
for col in df.columns[5:7]:
    df[col] = pd.to_datetime(df[col], format='%Y-%m-%dT%H:%M:%S.%f', errors='coerce')  
for col in df.columns[10:12]:
    df[col] = pd.to_datetime(df[col], format='%Y-%m-%dT%H:%M:%S.%f', errors='coerce')
```

```python
df['availability'] = df[
    ['codeshared_airline_name', 'codeshared_airline_iataCode', 'codeshared_airline_icaoCode',
     'codeshared_flight_number', 'codeshared_flight_iataNumber', 'codeshared_flight_icaoNumber']
].notnull().all(axis=1)
df['availability'] = df['availability'].replace({True: 'Available', False: 'Not Available'})
df
```

```python
df.isnull().sum()
```

```python
df['Days'] = df['departure_scheduledTime'].dt.day
df['Month'] = df['departure_scheduledTime'].dt.strftime('%b')
df['Year'] = df['departure_scheduledTime'].dt.year.astype(str)
df['Hour'] = df['departure_scheduledTime'].dt.hour
df = df.drop(columns=[
    'codeshared_airline_name', 
    'codeshared_airline_iataCode', 
    'codeshared_airline_icaoCode', 
    'codeshared_flight_number', 
    'codeshared_flight_iataNumber', 
    'codeshared_flight_icaoNumber'
])
df
```

```python
df.to_csv('Test.csv', index=False)
```
