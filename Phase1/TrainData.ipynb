```python
from docx import Document
import numpy as np
import pandas as pd 
import json
import re
import os
```

```python
def extractText(filename):
    doc = Document(filename)
    text = ""
    for paragraph in doc.paragraphs:
        text += paragraph.text.strip()
    cleaned_text = text.replace('\n', '').replace('\t', '')
    return json.loads(text)

def get_filenames(directory):
    filenames = os.listdir(directory)    
    return filenames
```

```python
def flatten_and_convert_to_dataframe(data):
    flattened_data = []
    for entry in data:
        flat_entry = {
            "Flight Number": entry["flight"].get("number", np.nan),
            "Type": entry["type"],
            "Status": entry["status"],
            "Departure IATA Code": entry["departure"].get("iataCode", np.nan),
            "Departure ICAO Code": entry["departure"].get("icaoCode", np.nan),
            "Scheduled Time": entry["departure"].get("scheduledTime", np.nan),
            "Actual Time": entry["departure"].get("actualTime", np.nan),
            "Arrival IATA Code": entry["arrival"].get("iataCode", np.nan),
            "Arrival ICAO Code": entry["arrival"].get("icaoCode", np.nan),
            "Arrival Estimated Time": entry["arrival"].get("estimatedTime", np.nan)
        }
        flattened_data.append(flat_entry)
    return pd.DataFrame(flattened_data)
```

```python
basePath = r'C:\Users\FARJAD FARHAT\Downloads\MLProject\ML-Proj-Dataset\Train'

names = get_filenames(basePath)
files = []
for i in names:
    print(i)
    out = extractText(os.path.join(basePath,i))
    files.append(out)
```

```python
dfs = [] 
for file in files :
    temp = flatten_and_convert_to_dataframe(file)
    dfs.append(temp)
```

```python
df = pd.concat(dfs,axis=0,ignore_index=True)
```

```python
columns = ['Scheduled Time','Actual Time','Arrival Estimated Time']
for col in columns:
    df[col] = pd.to_datetime(df[col], format='%Y-%m-%dT%H:%M:%S.%f', errors='coerce')
```

```python
df['Days'] = df['Scheduled Time'].dt.day
df['Month'] = df['Scheduled Time'].dt.strftime('%b')
df['Year'] = df['Scheduled Time'].dt.year.astype(str)
df['Hour'] = df['Scheduled Time'].dt.hour
df
```
