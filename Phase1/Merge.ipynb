```python
import pandas as pd 
import numpy as np
import os
from sklearn.preprocessing import StandardScaler
```

```python
def get_filenames(directory):
    filenames = os.listdir(directory)    
    return filenames
```

```python
basePath = r'C:\Users\FARJAD FARHAT\Downloads\MLProject\Phase1\Weather_Months'
names = get_filenames(basePath)
```

```python
dfs = []
for i in names:
    df = pd.read_csv(os.path.join(basePath,i))
    df['Month'] = i[0:3]
    if "2024" in i:
        df['Year'] = "2024"
    else:
        df['Year'] = "2023"
    dfs.append(df)
df = pd.concat(dfs,axis=0,ignore_index=True)
```

```python
scale = StandardScaler()
for col in df.columns[1:16]:
    df[col] = scale.fit_transform(df[col].values.reshape(-1,1))
df
```

```python
Traindf = pd.read_csv("Train.csv")
Traindf["Year"] = Traindf["Year"].astype(str)
```

```python
result = pd.merge(Traindf, df, on=["Days", "Month","Year"], how="inner", suffixes=("_df1", "_df2"))
```

```python
print(result.isnull().sum())
```

```python
result['Scheduled Time'] = pd.to_datetime(result['Scheduled Time'], errors='coerce')
result['Actual Time'] = pd.to_datetime(result['Actual Time'], errors='coerce')

result['departure_delay'] = np.where(
    result['Actual Time'].notna() & result['Scheduled Time'].notna(),
    (result['Actual Time'] - result['Scheduled Time']).dt.total_seconds() / 60, 
    np.nan  
)

result
```

```python
result.loc[result['departure_delay'] < 0, 'departure_delay'] = 0
result
```

```python
result['departure_delay'] = result['departure_delay'].fillna(result['departure_delay'].mean()).round(2)
result
```

```python
result.to_csv("NotScaledDelay.csv",index=False)
```
